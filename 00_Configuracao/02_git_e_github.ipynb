## 1. O problema que o Git resolve

Você já viveu isto:

```
monografia.docx
monografia_v2.docx
monografia_v2_revisado.docx
monografia_v2_revisado_FINAL.docx
monografia_FINAL_AGORAVAI.docx
monografia_FINAL_AGORAVAI (1).docx
```

Três perguntas que essa pasta não consegue responder:

1. O que exatamente mudou entre a `v2` e a `v2_revisado`?
2. Se eu quiser voltar para como estava há duas semanas, como faço?
3. Se duas pessoas editarem ao mesmo tempo, qual arquivo sobrevive?

**Git** é a ferramenta que responde as três. Ele é um sistema de **controle de versão**:
em vez de você criar cópias com nomes cada vez mais desesperados, ele guarda o histórico
do projeto inteiro e permite consultar, comparar e voltar a qualquer ponto do passado.

Ele foi criado em 2005 por Linus Torvalds — o mesmo do Linux — para coordenar milhares
de pessoas mexendo no mesmo código. Hoje é padrão absoluto: qualquer empresa de
tecnologia, qualquer banco com mesa quantitativa, qualquer laboratório de pesquisa usa
Git.

## 2. Git ≠ GitHub

Essa confusão é universal, então vale fixar:

| | **Git** | **GitHub** |
|---|---|---|
| O que é | Um programa que roda no seu computador | Um site que hospeda projetos Git |
| Precisa de internet? | Não | Sim |
| Analogia | O controle de alterações do Word | O Google Drive onde o arquivo mora |
| Alternativas | (praticamente nenhuma na prática) | GitLab, Bitbucket, Azure DevOps |

Uma frase para guardar: **Git guarda o histórico; GitHub guarda o Git na nuvem e coloca
uma interface web em cima.**

Você pode usar Git sozinho, offline, sem nunca criar conta em lugar nenhum. Mas na
prática as pessoas usam GitHub junto, porque é assim que se compartilha código, se
colabora e — muito importante para você — **se mostra trabalho**.

## 3. Vocabulário mínimo

Só cinco palavras. Você vai ouvir todas elas o tempo todo.

**Repositório** (ou *repo*) — a pasta do projeto, com todo o seu histórico junto. Este
material que você está lendo é um repositório.

**Commit** — uma "foto" do projeto num instante, com uma mensagem explicando o que
mudou. O histórico de um repositório é a sequência de commits. Pense em cada commit como
um ponto de salvamento de videogame.

**Clonar** (`clone`) — baixar um repositório inteiro, com histórico, para a sua máquina.
É o que você vai fazer para pegar o material.

**Puxar** (`pull`) — atualizar sua cópia local com o que mudou no GitHub. É o que você
vai fazer quando publicarmos uma aula nova.

**README** — o arquivo `README.md` na raiz do repositório. O GitHub o exibe
automaticamente na página inicial do projeto. É a porta de entrada: quem chega, lê isso.

E é só. *Branch*, *merge*, *pull request*, *fork* existem e são úteis, mas não hoje.

## 5. Lendo um repositório

Ao abrir um repositório no GitHub, você vê uma lista de arquivos e pastas e, abaixo
dela, o README renderizado. Vale saber olhar quatro coisas:

- **A lista de arquivos** — a estrutura do projeto. Pastas numeradas (`00_`, `01_`, ...)
  indicam ordem de leitura, como aqui.
- **A data ao lado de cada arquivo** — quando aquilo mudou pela última vez.
- **A aba `Commits`** — o histórico. Clicar em um commit mostra exatamente quais linhas
  foram adicionadas (verde) e removidas (vermelho).
- **A aba `Issues`** — onde se reportam problemas e dúvidas. Achou um erro no material?
  Abrir uma *issue* é o caminho certo, e é uma contribuição de verdade para o projeto.

Notebooks (`.ipynb`) o GitHub renderiza direto na página, com texto, código e gráficos.
Você consegue ler o material inteiro sem baixar nada.

Três caminhos, do mais simples ao mais completo. Escolha um.

### Caminho B — Download do ZIP

uma aula nova, você terá que baixar tudo de novo.

### Caminho C — Git no seu computador

Instale o Git ([git-scm.com/downloads](https://git-scm.com/downloads)) e, no terminal:

```bash

cd Introducao-a-Ciencia-de-Dados
```

Para atualizar depois, de dentro da pasta:

```bash
git pull
```

Esse `git pull` é a vantagem dos caminhos A e C sobre o B: um comando e você tem a
versão mais recente do material.

## 7. Confirmando que deu certo

Se você clonou o repositório e está com este notebook aberto de dentro dele, a célula
abaixo consegue enxergar a estrutura do projeto. Ela usa `pathlib`, que é o jeito
moderno de mexer com caminhos de arquivo em Python — não se preocupe com a sintaxe
agora, apenas rode.

````python
from pathlib import Path

raiz = Path("..")   # ".." significa "a pasta de cima", ou seja, a raiz do repositório

print("Conteúdo da raiz do repositório:\n")
for item in sorted(raiz.iterdir()):
    if item.name.startswith("."):
        continue                       # pula arquivos ocultos, como .git e .gitignore
    marcador = "[pasta]" if item.is_dir() else "       "
    print(f"  {marcador} {item.name}")
````

Você deve ver as pastas dos módulos (`00_` a `04_`), a pasta `data/`, a pasta `assets/`
e os arquivos `README.md` e `requirements.txt`.

Agora vamos confirmar que os dados que usaremos nas próximas semanas estão no lugar.
Todos os arquivos vêm **junto com o repositório** — o material nunca baixa dados da
internet durante a execução. Isso garante que tudo funcione mesmo com internet ruim, e
que os números das aulas não mudem no meio do semestre.

````python
pasta_dados = Path("../data")

for arquivo in sorted(pasta_dados.glob("*.*")):
    tamanho_kb = arquivo.stat().st_size / 1024
    print(f"  {arquivo.name:<28} {tamanho_kb:>8.1f} KB")
````

Rodou e listou os arquivos? Então seu ambiente está pronto e você pode ir para o
módulo 01.

Se apareceu `FileNotFoundError`, quase certamente é uma destas duas causas:

1. Você abriu o notebook **fora** da pasta do repositório (por exemplo, abriu só o
   arquivo `.ipynb` no Colab sem clonar o resto). Solução: use o Caminho A da seção 6.
2. Você está em uma pasta diferente da esperada. Rode `!pwd` (Linux/Mac/Colab) ou
   `!cd` (Windows) em uma célula para ver onde você está.
