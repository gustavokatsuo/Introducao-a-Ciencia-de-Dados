## Exercício 1 (teoria) — Ordem de execução


Uma colega abre um notebook, executa a célula 5, depois a célula 2, depois a 8. Ela diz
que o resultado final "está certo, porque todas as células rodaram".

Explique por que essa afirmação pode ser falsa e qual procedimento resolve a dúvida
definitivamente.

**Resposta:**


O notebook guarda **estado**: cada célula executada altera as variáveis que existem na
memória, e o que vale é a **ordem de execução**, não a ordem em que as células aparecem
na tela. Rodando 5 → 2 → 8, a célula 8 usa os valores deixados pela 2, que por sua vez
pode ter sobrescrito o que a 5 tinha criado. O resultado exibido pode depender de uma
variável que já não existe mais no arquivo, ou de um valor antigo.

O procedimento que resolve é **`Ambiente de execução → Reiniciar sessão e executar
tudo`**. Isso limpa a memória e roda o notebook de cima para baixo, na ordem em que ele
está escrito. Se rodar sem erro e reproduzir os mesmos resultados, o notebook está
correto; se quebrar, o que existia era uma ilusão sustentada pelo estado antigo.

## Exercício 4 (código) — Encontrando os dados


Os dados da capacitação estão na pasta `data/`, na raiz do repositório. Como este
notebook está dentro de `04_Exercicios/`, o caminho até lá é `../data`.

Escreva um código que liste o **nome** e o **tamanho em KB** de cada arquivo dessa pasta.


> **Dica:** `from pathlib import Path`; `Path('../data').glob('*.csv')` devolve os arquivos CSV, e `arquivo.stat().st_size` dá o tamanho em bytes.

````python
from pathlib import Path

pasta = Path("../data")

for arquivo in sorted(pasta.glob("*.*")):
    tamanho_kb = arquivo.stat().st_size / 1024
    print(f"{arquivo.name:<28} {tamanho_kb:>8.1f} KB")
````

## Exercício 5 (teoria) — Lendo um erro


Você executa uma célula e recebe:

```
Traceback (most recent call last):
  File "<ipython-input-4>", line 2, in <module>
    print(preco_medio)
NameError: name 'preco_medio' is not defined
```

Responda:

1. Qual linha do erro é a mais importante, e por quê?
2. Cite **duas** causas possíveis para esse erro específico.
3. O que você digitaria em uma busca para pedir ajuda?

**Resposta:**


1. A **última linha** — `NameError: name 'preco_medio' is not defined`. Ela traz o
   **tipo** do erro e a **explicação**. As linhas de cima são o rastro de onde o erro
   aconteceu, útil em programas grandes; em um notebook, a última linha basta na maioria
   dos casos.

2. Duas causas possíveis:
   - **erro de digitação** — a variável foi criada como `preco_médio` ou `precos_medio`;
   - **a célula que cria a variável não foi executada** (ou foi executada e depois a
     sessão reiniciou, apagando a memória).

3. Busque pela **mensagem do erro**, não pela sua descrição dela:
   `python NameError name is not defined`. Ao pedir ajuda a uma pessoa, envie a última
   linha do erro **e** a célula que você rodou — "deu erro" sozinho não é uma pergunta
   respondível.

## Exercício 6 (código) — Provocando e corrigindo um erro


Faça duas coisas em uma única célula:

1. provoque um `NameError` de propósito, capturando-o com `try` / `except` e imprimindo
   a mensagem;
2. logo abaixo, corrija o problema e imprima o valor correto.


> **Dica:** A estrutura é `try:` / `except NameError as erro:` — e você já viu esse padrão no notebook 01 do módulo 00.

````python
# 1. provocando o erro
try:
    print(taxa_selic)
except NameError as erro:
    print("NameError capturado:", erro)

# 2. corrigindo: basta criar a variável antes de usá-la
taxa_selic = 0.1075
print("Agora funciona:", taxa_selic)
````

`try` / `except` serve para lidar com erros previstos sem interromper o programa. Você
não vai usá-lo o tempo todo — no dia a dia da análise, é melhor deixar o erro aparecer e
corrigir a causa. Aqui ele serve para demonstrar o erro sem travar o notebook.

## Exercício 7 (teoria) — Escolhendo como baixar o material


Existem três formas de obter o material: (A) `!git clone` no Colab, (B) download do ZIP
pelo site do GitHub, (C) `git clone` no terminal do seu computador.

Uma colega vai acompanhar a capacitação inteira pelo Colab, em um notebook emprestado
onde ela não pode instalar nada. Qual você recomenda, e por quê? Qual é a principal
desvantagem da opção (B)?

**Resposta:**


Recomenda-se a opção **(A)**: `!git clone` dentro do próprio Colab. Ela não exige
instalar nada na máquina (tudo roda nos servidores do Google), traz o repositório
completo — inclusive a pasta `data/` — e faz os caminhos usados no material
(`../data/...`) funcionarem sem alteração.

A desvantagem da opção **(B)** é que o ZIP é uma **fotografia** do repositório: ele não
se atualiza. Quando o material for corrigido ou ganhar uma aula nova, será preciso
baixar tudo de novo e reorganizar os arquivos manualmente. Com Git (A ou C), basta um
`git pull`.
