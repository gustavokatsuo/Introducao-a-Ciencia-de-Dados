## 1. Listas

A estrutura mais usada de todas. Uma sequência ordenada de itens, entre colchetes.

````python
carteira = ["PETR4", "VALE3", "ITUB4", "WEGE3"]
precos = [32.50, 61.20, 28.75, 41.90]
mista = ["PETR4", 32.50, 100, True]    # pode misturar tipos, embora raramente convenha

print(carteira)
print(len(carteira))
print(type(carteira))
````

### Acessando itens

Índices começam em **0**. Negativos contam de trás para frente.

````python
print(carteira[0])     # primeiro
print(carteira[2])     # terceiro
print(carteira[-1])    # último
print(carteira[-2])    # penúltimo
````

````python
# Fatiamento: [inicio:fim], inclui o início e exclui o fim
print(carteira[0:2])
print(carteira[:2])     # do começo até o índice 2 (exclusive)
print(carteira[2:])     # do índice 2 até o fim
print(carteira[::2])    # de dois em dois
print(carteira[::-1])   # invertida
````

````python
# Índice inexistente gera erro:
try:
    print(carteira[10])
except IndexError as erro:
    print("IndexError:", erro)
````

### Modificando listas

Listas são **mutáveis**: dá para trocar, adicionar e remover itens depois de criadas.

````python
carteira = ["PETR4", "VALE3", "ITUB4"]

carteira[1] = "VALE5"              # troca no lugar
print(carteira)

carteira.append("WEGE3")           # adiciona no fim
print(carteira)

carteira.insert(0, "ABEV3")        # adiciona numa posição
print(carteira)

carteira.remove("VALE5")           # remove pelo valor
print(carteira)

ultimo = carteira.pop()            # remove o último E devolve o valor
print(ultimo, "->", carteira)
````

````python
# Juntar e ordenar
carteira_a = ["PETR4", "VALE3"]
carteira_b = ["ITUB4", "WEGE3"]

completa = carteira_a + carteira_b       # concatena, cria uma lista nova
print(completa)

completa.extend(["ABEV3"])               # adiciona vários itens no fim
print(completa)

print(sorted(completa))                  # devolve uma cópia ordenada
print(completa)                          # a original continua intacta

completa.sort()                          # ordena a original, no lugar
print(completa)
````

> **`sorted()` vs `.sort()`** — essa distinção vale para várias operações em Python:
> `sorted(lista)` devolve uma **nova** lista ordenada e não mexe na original;
> `lista.sort()` **modifica** a original e devolve `None`.
>
> Um erro comum de iniciante é escrever `carteira = carteira.sort()` e acabar com
> `None` na variável.

````python
numeros = [3, 1, 2]
resultado = numeros.sort()
print(resultado)      # None! .sort() não devolve a lista
print(numeros)        # a lista foi ordenada no lugar
````

### Funções úteis com listas de números

````python
precos = [32.50, 61.20, 28.75, 41.90, 15.30]

print("Quantidade:", len(precos))
print("Soma:", sum(precos))
print("Máximo:", max(precos))
print("Mínimo:", min(precos))
print("Média:", sum(precos) / len(precos))
print("PETR4 está na lista?", 32.50 in precos)     # o operador "in" testa pertencimento
````

## 2. Compreensões de lista (*list comprehensions*)

Um padrão aparece o tempo todo: percorrer uma lista e construir outra a partir dela.

```python
nova = []
for item in original:
    nova.append(transformacao(item))
```

Python tem uma sintaxe compacta para exatamente isso:

```python
nova = [transformacao(item) for item in original]
```

Leia da esquerda para a direita: *"a transformação de cada item, para cada item na
original"*.

````python
precos_reais = [32.50, 61.20, 28.75, 41.90]
cotacao_dolar = 5.45

# jeito longo:
precos_dolar = []
for preco in precos_reais:
    precos_dolar.append(round(preco / cotacao_dolar, 2))
print(precos_dolar)

# compreensão de lista:
precos_dolar = [round(preco / cotacao_dolar, 2) for preco in precos_reais]
print(precos_dolar)
````

Você também pode **filtrar**, acrescentando um `if` no fim:

```python
nova = [item for item in original if condicao]
```

````python
precos = [32.50, 61.20, 28.75, 41.90, 15.30]

baratos = [preco for preco in precos if preco < 35]
print(baratos)

# transformando e filtrando ao mesmo tempo:
tickers = ["petr4", "vale3", "itub4", "wege3"]
maiusculos = [t.upper() for t in tickers if t.endswith("3")]
print(maiusculos)
````

> **Quando usar?** Quando a operação cabe confortavelmente em uma linha e continua
> legível. Se você precisar de dois `if`, um `else` e uma conta complicada, escreva o
> `for` normal — código curto que ninguém entende não é código bom.

## 3. Dicionários

Listas guardam itens por **posição**. Dicionários guardam por **nome** — o que é muito
mais natural quando cada valor tem um significado.

A sintaxe é `{chave: valor}`.

````python
acao = {
    "ticker": "PETR4",
    "empresa": "Petrobras",
    "setor": "Petróleo e Gás",
    "preco": 32.50,
    "dividend_yield": 0.147,
}

print(acao)
print(acao["ticker"])
print(acao["preco"])
````

Compare com a versão em lista:

```python
acao = ["PETR4", "Petrobras", "Petróleo e Gás", 32.50, 0.147]
acao[3]   # ... o que era mesmo o índice 3?
```

`acao["preco"]` se explica sozinho. É por isso que dicionários são onipresentes.

````python
# Acessar chave inexistente gera erro:
try:
    print(acao["p_l"])
except KeyError as erro:
    print("KeyError:", erro)

# .get() evita o erro e permite um valor padrão:
print(acao.get("p_l"))                    # None
print(acao.get("p_l", "não informado"))   # valor padrão
````

````python
# Adicionar, alterar e remover
acao["p_l"] = 4.2              # chave nova: adiciona
acao["preco"] = 33.10          # chave existente: sobrescreve
print(acao)

del acao["dividend_yield"]
print(acao.keys())
````

### Percorrendo um dicionário

````python
precos = {"PETR4": 32.50, "VALE3": 61.20, "ITUB4": 28.75, "WEGE3": 41.90}

print("--- só as chaves ---")
for ticker in precos.keys():
    print(ticker, end=" ")

print("\n\n--- só os valores ---")
for preco in precos.values():
    print(preco, end=" ")

print("\n\n--- os dois juntos (o mais usado) ---")
for ticker, preco in precos.items():
    print(f"{ticker}: R$ {preco:.2f}")
````

````python
# "in" testa a existência de uma CHAVE:
print("PETR4" in precos)
print(32.50 in precos)            # False: 32.50 é valor, não chave
print(32.50 in precos.values())   # True
````

### Dicionários aninhados e listas de dicionários

Estruturas reais quase sempre são combinações. Guarde bem o padrão abaixo — **lista de
dicionários** é exatamente a forma que a maioria das APIs devolve dados, e é a ponte
direta para o `DataFrame` do pandas, no módulo 02.

````python
carteira = [
    {"ticker": "PETR4", "preco": 32.50, "quantidade": 100, "setor": "Petróleo e Gás"},
    {"ticker": "VALE3", "preco": 61.20, "quantidade": 50,  "setor": "Mineração"},
    {"ticker": "ITUB4", "preco": 28.75, "quantidade": 200, "setor": "Financeiro"},
    {"ticker": "WEGE3", "preco": 41.90, "quantidade": 80,  "setor": "Bens industriais"},
]

valor_total = 0
for posicao in carteira:
    valor = posicao["preco"] * posicao["quantidade"]
    valor_total += valor
    print(f"{posicao['ticker']:<8} R$ {valor:>10,.2f}")

print(f"{'TOTAL':<8} R$ {valor_total:>10,.2f}")
````

````python
# Agrupando por setor — a versão "na mão" do groupby que veremos em pandas:
valor_por_setor = {}

for posicao in carteira:
    setor = posicao["setor"]
    valor = posicao["preco"] * posicao["quantidade"]
    valor_por_setor[setor] = valor_por_setor.get(setor, 0) + valor

for setor, valor in valor_por_setor.items():
    participacao = valor / valor_total
    print(f"{setor:<20} R$ {valor:>10,.2f}   ({participacao:.1%})")
````

> Guarde essa célula. Em duas semanas você vai fazer exatamente isso com
> `df.groupby("setor")["valor"].sum()`, em uma linha. Ter feito na mão uma vez é o que
> faz a linha do pandas parar de ser mágica e passar a ser óbvia.

## 4. Tuplas

Uma tupla é uma lista que **não pode ser alterada** depois de criada. Usa parênteses.

````python
coordenada = (32.50, 61.20)
print(coordenada)
print(coordenada[0])

try:
    coordenada[0] = 40.00
except TypeError as erro:
    print("TypeError:", erro)
````

"Para que serve algo que não posso mudar?" Três respostas práticas:

1. **Proteção.** Dados que não devem mudar (uma data, um par fixo de valores) ficam
   protegidos de alteração acidental.
2. **Podem ser chave de dicionário.** Listas não podem; tuplas sim.
3. **Você já usa sem saber.** Funções que devolvem vários valores devolvem tuplas, e
   `.items()` de um dicionário entrega tuplas.

````python
# Desempacotamento: distribuir os elementos em variáveis
minimo, maximo = (28.75, 61.20)
print(minimo, maximo)

# É isso que acontece silenciosamente aqui:
precos = {"PETR4": 32.50, "VALE3": 61.20}
for par in precos.items():
    print(par, type(par))
````

## 5. Conjuntos (*sets*)

Um conjunto é uma coleção **sem ordem** e **sem repetições** — o conjunto da matemática.
Usa chaves, como o dicionário, mas sem `chave: valor`.

````python
setores = {"Financeiro", "Mineração", "Financeiro", "Petróleo e Gás", "Financeiro"}
print(setores)          # as repetições somem
print(len(setores))
````

O uso mais frequente no dia a dia é **descobrir os valores distintos** de uma lista:

````python
registros = ["PETR4", "VALE3", "PETR4", "ITUB4", "VALE3", "PETR4"]

distintos = set(registros)
print(distintos)
print(f"{len(registros)} registros, {len(distintos)} ativos distintos")
````

````python
# Operações de conjunto — úteis para comparar duas carteiras:
carteira_2024 = {"PETR4", "VALE3", "ITUB4", "MGLU3"}
carteira_2025 = {"PETR4", "ITUB4", "WEGE3", "ABEV3"}

print("Mantidos:      ", carteira_2024 & carteira_2025)   # interseção
print("Todos já vistos:", carteira_2024 | carteira_2025)   # união
print("Vendidos:      ", carteira_2024 - carteira_2025)   # diferença
print("Comprados:     ", carteira_2025 - carteira_2024)
````

## 6. Escolhendo a estrutura certa

| Sua pergunta | Estrutura |
|---|---|
| "Preciso de uma sequência de itens que posso alterar" | **lista** |
| "Cada valor tem um nome/rótulo" | **dicionário** |
| "Isso não pode mudar" | **tupla** |
| "Quais são os valores únicos?" / "o que há em comum entre A e B?" | **conjunto** |

E um alerta que vale ouro daqui a duas semanas: **para tabelas de dados, nenhuma delas.**
Aí você usa o `DataFrame` do pandas. Estas quatro são a fundação — as peças que o pandas
usa por baixo e que você continuará usando para tudo que não for tabela.

## 7. Uma armadilha: listas são referências

Isto pega todo mundo pelo menos uma vez.

````python
original = ["PETR4", "VALE3"]
copia = original          # isto NÃO cria uma cópia!

copia.append("ITUB4")

print("copia   :", copia)
print("original:", original)     # a original também mudou
````

`copia = original` não copia nada: cria um **segundo nome para a mesma lista**. Pense em
dois atalhos apontando para o mesmo arquivo.

Para copiar de verdade, use `.copy()`:

````python
original = ["PETR4", "VALE3"]
copia = original.copy()

copia.append("ITUB4")

print("copia   :", copia)
print("original:", original)     # intacta
````

Isso vale para listas, dicionários e conjuntos (tudo que é mutável), e volta a aparecer
em pandas — onde a mensagem `SettingWithCopyWarning` tem exatamente essa raiz.

## 8. Recapitulando

- **Lista** `[ ]`: ordenada e alterável. `append`, `insert`, `remove`, `pop`, `sort`.
- **Compreensão de lista** `[f(x) for x in lista if cond]`: transforma e filtra em uma
  linha — desde que continue legível.
- **Dicionário** `{chave: valor}`: acesso por nome. `.get()` para não quebrar,
  `.items()` para percorrer.
- **Tupla** `( )`: imutável. Aparece em desempacotamento e em retornos de função.
- **Conjunto** `{ }`: sem ordem e sem repetição. Ideal para valores únicos e comparações.
- `sorted()` devolve cópia; `.sort()` altera no lugar e devolve `None`.
- Atribuir uma lista a outra variável **não copia** — use `.copy()`.
