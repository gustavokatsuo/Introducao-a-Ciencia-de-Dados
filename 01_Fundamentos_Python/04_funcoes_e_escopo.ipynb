## 1. Por que funções?

Olhe este código, que calcula o retorno de três ativos:

```python
retorno_petr = (32.45 - 30.10) / 30.10
print(f"PETR4: {retorno_petr:.2%}")

retorno_vale = (61.20 - 68.40) / 68.40
print(f"VALE3: {retorno_vale:.2%}")

retorno_itub = (28.75 - 26.90) / 26.90
print(f"ITUB4: {retorno_itub:.2%}")
```

A mesma conta, escrita três vezes. Se a fórmula mudar — digamos, para descontar custos —
você precisa lembrar de alterar nos três lugares. E vai esquecer de um.

Funções resolvem isso empacotando um trecho de código sob um nome. Os ganhos:

- **Não repetir.** A regra fica escrita em um lugar só.
- **Nomear.** `calcular_retorno(...)` diz o que faz; a fórmula solta, não.
- **Testar.** Uma função isolada pode ser conferida com casos conhecidos.
- **Reaproveitar.** Escreveu bem uma vez, usa em todos os projetos.

## 2. Definindo uma função

```python
def nome_da_funcao(parametro1, parametro2):
    """Docstring: o que a função faz."""
    # corpo, indentado
    return resultado
```

Repare nos elementos obrigatórios: a palavra `def`, os parênteses, os dois-pontos e a
indentação do corpo — as mesmas regras do `if` e do `for`.

````python
def calcular_retorno(preco_inicial, preco_final):
    """Calcula o retorno percentual entre dois preços."""
    return (preco_final - preco_inicial) / preco_inicial
````

Executar a célula acima **não calculou nada**: apenas ensinou o Python a fazer aquilo.
Para usar, é preciso **chamar** a função.

````python
retorno = calcular_retorno(30.10, 32.45)
print(retorno)
print(f"{retorno:.2%}")
````

````python
# Agora os três ativos, sem repetir a fórmula:
ativos = ["PETR4", "VALE3", "ITUB4"]
compras = [30.10, 68.40, 26.90]
atuais = [32.45, 61.20, 28.75]

for ticker, compra, atual in zip(ativos, compras, atuais):
    print(f"{ticker}: {calcular_retorno(compra, atual):.2%}")
````

### Parâmetro × argumento

Duas palavras que as pessoas usam como sinônimo, mas têm significados distintos:

- **parâmetro** é o nome na definição (`preco_inicial`, `preco_final`);
- **argumento** é o valor passado na chamada (`30.10`, `32.45`).

Por padrão, os argumentos são associados **pela ordem**. Você também pode nomeá-los na
chamada, o que deixa o código muito mais claro quando há vários parâmetros:

````python
print(calcular_retorno(30.10, 32.45))                                  # por posição
print(calcular_retorno(preco_inicial=30.10, preco_final=32.45))        # por nome
print(calcular_retorno(preco_final=32.45, preco_inicial=30.10))        # nome: a ordem não importa
````

## 3. `return` × `print` — leia esta seção duas vezes

Esta confusão trava mais gente do que qualquer outro conceito inicial.

| | `print` | `return` |
|---|---|---|
| O que faz | **mostra** um texto na tela | **devolve** um valor para quem chamou |
| Para quem | para o ser humano que está olhando | para o resto do programa |
| Dá para guardar o resultado? | não | sim |
| Encerra a função? | não | sim, imediatamente |

Vamos ver as duas versões lado a lado.

````python
def retorno_com_return(preco_inicial, preco_final):
    return (preco_final - preco_inicial) / preco_inicial


def retorno_com_print(preco_inicial, preco_final):
    print((preco_final - preco_inicial) / preco_inicial)
````

````python
a = retorno_com_return(30.10, 32.45)
b = retorno_com_print(30.10, 32.45)     # a função já imprimiu, por conta própria

print("valor guardado de a:", a)
print("valor guardado de b:", b)        # None! a função não devolveu nada
````

Aí está: `retorno_com_print` mostrou o número, mas **não entregou nada**. A variável `b`
ficou com `None`, e qualquer tentativa de usá-la quebra:

````python
try:
    print(b * 100)
except TypeError as erro:
    print("TypeError:", erro)

print(a * 100)     # com return, o valor é utilizável
````

**A regra prática:** uma função que **calcula** algo deve usar `return`. Se você quiser
ver o resultado, dê `print` **no que ela devolveu** — a decisão de exibir é de quem
chama, não da função.

```python
print(calcular_retorno(30.10, 32.45))   # certo: a função calcula, quem chama exibe
```

Funções que existem só para exibir (montar um relatório na tela, por exemplo) podem usar
`print` — mas são a exceção.

> **Dica:** No notebook há uma pegadinha visual: se a **última linha** de uma célula é uma
> expressão, o Jupyter mostra o valor automaticamente, sem `print`. Isso pode dar a
> falsa impressão de que a função "imprimiu" algo. Ela não imprimiu; foi o notebook.

````python
calcular_retorno(30.10, 32.45)     # sem print, mas o notebook exibe o valor da última linha
````

### `return` encerra a função na hora

````python
def classificar_risco(volatilidade):
    if volatilidade > 0.40:
        return "Alto"
    if volatilidade > 0.20:
        return "Médio"
    return "Baixo"
    print("esta linha nunca será executada")


print(classificar_risco(0.55))
print(classificar_risco(0.30))
print(classificar_risco(0.12))
````

## 4. Valores padrão

Um parâmetro pode ter valor padrão, usado quando o argumento não é informado. Isso
deixa a função flexível sem obrigar quem chama a decidir tudo.

````python
def calcular_montante(capital, taxa, periodos=12):
    """Montante de juros compostos. Por padrão, 12 períodos."""
    return capital * (1 + taxa) ** periodos


print(round(calcular_montante(10_000, 0.008), 2))          # usa o padrão: 12
print(round(calcular_montante(10_000, 0.008, 24), 2))      # sobrescreve
print(round(calcular_montante(10_000, 0.008, periodos=60), 2))
````

> **Atenção:** Parâmetros com valor padrão vêm **depois** dos sem valor padrão. `def
> f(a=1, b)` é erro de sintaxe.

## 5. Docstrings

A *docstring* é o texto entre aspas triplas logo abaixo do `def`. Não é enfeite: é como
o Python documenta funções, e é o que aparece quando alguém pede ajuda sobre a sua.

Uma boa docstring responde três perguntas: **o que faz**, **o que recebe**, **o que
devolve**.

````python
def calcular_retorno_anualizado(retorno_total, anos):
    """
    Converte um retorno acumulado em retorno anualizado (média geométrica).

    Parâmetros
    ----------
    retorno_total : float
        Retorno acumulado no período, em decimal (0.35 = 35%).
    anos : float
        Duração do período, em anos.

    Retorna
    -------
    float
        Retorno anualizado equivalente, em decimal.

    Exemplo
    -------
    >>> calcular_retorno_anualizado(0.35, 3)
    0.1052...
    """
    return (1 + retorno_total) ** (1 / anos) - 1


print(f"{calcular_retorno_anualizado(0.35, 3):.2%} ao ano")
````

````python
help(calcular_retorno_anualizado)
````

Repare que `help()` funciona exatamente igual ao que você usaria com uma função de
biblioteca — `help(pd.read_csv)`, por exemplo. Documentar não é burocracia; é o que
torna o seu código consultável.

## 6. Devolvendo vários valores

Basta separar por vírgula no `return`. Por baixo, o Python empacota tudo em uma tupla —
e você pode desempacotar na chamada.

````python
def resumir(valores):
    """Devolve mínimo, máximo e média de uma lista de números."""
    return min(valores), max(valores), sum(valores) / len(valores)


precos = [32.50, 61.20, 28.75, 41.90, 15.30]

minimo, maximo, media = resumir(precos)

print(f"Mínimo: {minimo:.2f} | Máximo: {maximo:.2f} | Média: {media:.2f}")
````

## 7. Escopo: onde cada variável existe

Variáveis criadas **dentro** de uma função existem apenas ali. Elas nascem quando a
função é chamada e somem quando ela termina. Esse território privado é o **escopo local**.

Variáveis criadas fora de qualquer função vivem no **escopo global** e podem ser lidas
de qualquer lugar.

````python
taxa_cdi = 0.1075          # global


def calcular_rendimento(capital):
    meses = 12             # LOCAL: só existe aqui dentro
    return capital * taxa_cdi * (meses / 12)   # consegue LER a global


print(calcular_rendimento(10_000))

try:
    print(meses)
except NameError as erro:
    print("NameError:", erro)
````

### Ler é fácil; escrever é outra história

Uma função consegue **ler** uma variável global sem problema. Mas se ela **atribuir** um
valor a um nome, o Python cria uma variável local nova — a global fica intacta.

````python
contador = 0


def incrementar_errado():
    contador = 100          # cria uma variável LOCAL chamada contador
    return contador


print("dentro :", incrementar_errado())
print("fora   :", contador)     # a global não mudou
````

Existe a palavra-chave `global` para forçar a alteração da variável externa. Nós a
mostramos aqui para você **reconhecê-la**, não para usá-la:

````python
contador = 0


def incrementar_com_global():
    global contador
    contador += 1


incrementar_com_global()
incrementar_com_global()
print(contador)
````

> **Atenção — Evite `global`.** Uma função que altera o mundo lá fora é imprevisível: você
> lê a chamada e não faz ideia do que ela mexeu. O padrão saudável é *receber por
> parâmetro, devolver por `return`*:

```python
def incrementar(valor):
    return valor + 1

contador = incrementar(contador)   # explícito: dá para ver o que mudou
```

Funções assim — que só dependem do que recebem e só comunicam pelo `return` — são
fáceis de testar, de reaproveitar e de entender. Busque escrever assim.

## 8. `lambda`: função de uma linha

`lambda` cria uma função pequena, sem nome, para usar na hora:

```python
lambda parametros: expressao
```

Ela existe para aqueles casos em que você precisa passar uma função como argumento para
outra função. Você vai reencontrá-la em pandas (`.apply()`, `.sort_values(key=...)`), e
é por isso que ela aparece aqui.

````python
dobro = lambda x: x * 2          # equivalente a: def dobro(x): return x * 2
print(dobro(21))

# O uso de verdade: como argumento de outra função
carteira = [
    {"ticker": "PETR4", "valor": 3250.0},
    {"ticker": "VALE3", "valor": 3060.0},
    {"ticker": "ITUB4", "valor": 5750.0},
]

por_valor = sorted(carteira, key=lambda posicao: posicao["valor"], reverse=True)

for posicao in por_valor:
    print(f"{posicao['ticker']}: R$ {posicao['valor']:,.2f}")
````

> Use `lambda` só para expressões curtas passadas como argumento. Se a lógica tem mais
> de uma linha ou vai ser usada em vários lugares, escreva um `def` com nome. Aliás,
> `dobro = lambda x: x * 2` é justamente o caso em que **não** se deve usar `lambda`:
> se a função tem nome, use `def`.

## 9. Juntando tudo

Uma pequena "biblioteca" de análise de carteira, escrita só com o que vimos no módulo 01.

````python
def calcular_valor_posicao(preco, quantidade):
    """Valor de mercado de uma posição."""
    return preco * quantidade


def calcular_retorno(preco_inicial, preco_final):
    """Retorno percentual entre dois preços, em decimal."""
    return (preco_final - preco_inicial) / preco_inicial


def classificar_posicao(retorno, limite=0.10):
    """Classifica uma posição pelo retorno, usando um limite de relevância."""
    if retorno > limite:
        return "Lucro relevante"
    if retorno > 0:
        return "Lucro pequeno"
    if retorno > -limite:
        return "Perda pequena"
    return "Perda relevante"


def analisar_carteira(posicoes):
    """
    Analisa uma carteira e devolve (relatório, valor total).

    Parâmetros
    ----------
    posicoes : list[dict]
        Cada dicionário precisa ter: ticker, preco_compra, preco_atual, quantidade.

    Retorna
    -------
    tuple[list[dict], float]
        Lista de dicionários com a análise de cada posição, e o valor total investido.
    """
    relatorio = []
    valor_total = 0.0

    for posicao in posicoes:
        valor = calcular_valor_posicao(posicao["preco_atual"], posicao["quantidade"])
        retorno = calcular_retorno(posicao["preco_compra"], posicao["preco_atual"])
        valor_total += valor

        relatorio.append({
            "ticker": posicao["ticker"],
            "valor": valor,
            "retorno": retorno,
            "situacao": classificar_posicao(retorno),
        })

    return relatorio, valor_total
````

````python
carteira = [
    {"ticker": "PETR4", "preco_compra": 30.10, "preco_atual": 32.45, "quantidade": 100},
    {"ticker": "VALE3", "preco_compra": 68.40, "preco_atual": 61.20, "quantidade": 50},
    {"ticker": "ITUB4", "preco_compra": 26.90, "preco_atual": 28.75, "quantidade": 200},
    {"ticker": "MGLU3", "preco_compra": 15.60, "preco_atual": 9.85,  "quantidade": 300},
]

relatorio, total = analisar_carteira(carteira)

print(f"{'ATIVO':<8}{'VALOR':>12}{'PESO':>9}{'RETORNO':>10}   SITUAÇÃO")
print("-" * 60)
for linha in relatorio:
    peso = linha["valor"] / total
    print(f"{linha['ticker']:<8}{linha['valor']:>12,.2f}{peso:>9.1%}"
          f"{linha['retorno']:>10.2%}   {linha['situacao']}")
print("-" * 60)
print(f"{'TOTAL':<8}{total:>12,.2f}")
````

## 10. Recapitulando

- `def nome(parametros):` define uma função; ela só executa quando é **chamada**.
- **`return` devolve um valor** para quem chamou; **`print` só mostra na tela**. Função
  que calcula usa `return`. Quem chama decide se exibe.
- `return` encerra a função imediatamente.
- Parâmetros podem ter **valores padrão**, e argumentos podem ser passados **por nome**.
- **Docstrings** documentam o que a função faz, recebe e devolve; `help()` as exibe.
- Variáveis criadas dentro da função são **locais** e somem ao final. Evite `global`:
  receba por parâmetro, devolva por `return`.
- `lambda` é uma função de uma linha, útil como argumento de outra função.
