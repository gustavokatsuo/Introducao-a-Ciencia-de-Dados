## 1. Por que notebooks?

Um programa tradicional é um arquivo de texto que roda do começo ao fim e cospe um
resultado. Isso funciona bem para construir um sistema, mas é péssimo para **investigar
dados**, que é o que nós vamos fazer.

Investigar dados é uma conversa: você carrega uma planilha, olha, estranha um número,
filtra, faz um gráfico, muda de ideia, tenta outra coisa. É um processo de idas e vindas.

O **notebook** foi feito exatamente para isso. Ele é um documento dividido em pedaços
(as *células*) que você executa na ordem que quiser, vendo o resultado de cada pedaço
logo abaixo dele. E ele mistura três coisas no mesmo arquivo:

| O que | Para que serve |
|---|---|
| **Texto** (como este) | Explicar o raciocínio, documentar decisões |
| **Código** | Fazer a conta, o filtro, o gráfico |
| **Resultado** | Tabelas, números e gráficos gerados pelo código |

É por isso que o notebook virou o formato padrão da área: o resultado de uma análise não
é só o número final, é o **caminho** até ele. E o caminho precisa estar visível para
alguém poder conferir.

O arquivo tem extensão `.ipynb` e é sempre o mesmo, não importa onde você o abra: o
mesmo notebook roda no Colab, no VS Code ou no JupyterLab.

## 2. Células

Este documento é feito de células. Existem dois tipos:

**Células de texto** (*markdown*) — é o que você está lendo agora. Servem para escrever
em português: títulos, explicações, conclusões.

**Células de código** — contêm Python. Você as executa e o resultado aparece logo abaixo.

A célula seguinte é de código. Clique nela e aperte `Shift + Enter`.

````python
print("Olá, FEA.dev!")
````

Executou? Apareceu `Olá, FEA.dev!` embaixo da célula.

Três atalhos que você vai usar o tempo todo:

| Atalho | O que faz |
|---|---|
| `Shift + Enter` | Executa a célula e vai para a próxima |
| `Ctrl + Enter` | Executa a célula e fica nela (bom para ir ajustando) |
| `Alt + Enter` | Executa e cria uma célula nova embaixo |

E dois detalhes visuais importantes, que funcionam igual no Colab e no VS Code:

- Enquanto uma célula roda, aparece uma **bolinha girando** à esquerda dela.
- Depois que roda, aparece um **número** entre colchetes: `[1]`, `[2]`, `[3]`...
  Esse número é a **ordem em que você executou**, não a ordem em que a célula aparece
  na tela. Guarde essa informação — ela vai ser importante daqui a pouco.

## 3. O notebook tem memória

Esta é a ideia mais importante da aula.

Quando você executa uma célula, tudo que ela criou **continua existindo** para as
próximas células. O notebook mantém uma memória viva do que já foi executado — o que
chamamos de **estado**.

Execute as duas células abaixo, na ordem.

````python
capital_inicial = 1000
````

````python
# Esta célula usa uma variável que foi criada na célula ANTERIOR.
print(capital_inicial * 1.10)
````

A segunda célula não tem nenhuma menção a `1000`, mas mesmo assim funcionou: ela
"lembrou" do que a primeira célula criou.

Isso é ótimo — é o que permite carregar uma base de dados grande uma vez e ficar
trabalhando em cima dela. Mas gera a armadilha clássica de quem começa:

> **O notebook não roda de cima para baixo sozinho. Ele roda na ordem em que VOCÊ
> clica.**

Se você executar a célula 5, depois voltar e executar a célula 2, o estado do notebook
reflete essa ordem maluca — e não o que está escrito na tela. É assim que surgem os
famosos "mas na minha máquina funcionava".

**A regra de ouro:** antes de entregar ou compartilhar qualquer notebook, reinicie e
rode tudo do zero.

| Ambiente | Como fazer |
|---|---|
| Colab | `Ambiente de execução → Reiniciar sessão e executar tudo` |
| VS Code | Botões `Reiniciar` e `Executar tudo`, na barra de cima do notebook |

Se ele rodar limpo de cima para baixo, ele está correto. Se quebrar, o que você tinha
era uma ilusão sustentada pelo estado antigo.

````python
# Uma demonstração do estado em ação:
capital_inicial = capital_inicial * 2   # usa o valor atual e sobrescreve
print(capital_inicial)
````

Execute a célula acima **três vezes seguidas** e veja o valor dobrar a cada execução:
2000, 4000, 8000.

O código é o mesmo; o resultado muda. O que mudou foi o estado. Reexecutar uma célula
nem sempre é inofensivo.

### Passo 5 — Criar o ambiente virtual e instalar as dependências

Um **ambiente virtual** (`venv`) é uma caixinha isolada com as bibliotecas de um projeto
específico. Ele evita que a versão de pandas que este material precisa brigue com a versão
que outro projeto seu precisa. Não é obrigatório, mas é um hábito barato de adquirir e
resolve problemas que são bem chatos de diagnosticar depois.

Abra o terminal **dentro do VS Code**: menu `Terminal → Novo Terminal`, ou
`` Ctrl + ` ``. Ele já abre na pasta do repositório, que é onde os comandos precisam
rodar.

**1. Criar o ambiente:**

```bash
python -m venv .venv        # Windows
python3 -m venv .venv       # macOS e Linux
```

Isso cria uma pasta `.venv/` dentro do repositório. Ela já está no `.gitignore`, então não
vai parar no GitHub.

**2. Ativar o ambiente:**

```bash
.venv\Scripts\activate        # Windows (PowerShell ou Prompt de Comando)
source .venv/bin/activate     # macOS e Linux
```

Deu certo quando o seu terminal passa a mostrar `(.venv)` no começo da linha.

> **PowerShell reclamando de "execução de scripts desabilitada"?** É uma trava padrão do
> Windows. Rode `Set-ExecutionPolicy -Scope CurrentUser RemoteSigned`, confirme com `S` (ou `Y`,
se o Windows estiver em inglês), e
> tente ativar de novo.

**3. Instalar as bibliotecas do material:**

```bash
pip install -r requirements.txt
```

O arquivo [`requirements.txt`](../requirements.txt) na raiz do repositório lista tudo que
o material usa — `numpy`, `pandas`, `matplotlib`, `seaborn`, `scipy`, `openpyxl` e o
próprio Jupyter. A instalação demora alguns minutos na primeira vez.

## 8. Erros: você vai ver muitos, e está tudo bem

Ver mensagem de erro em vermelho não significa que você é ruim nisso. Significa que
você está programando. Profissionais experientes veem erros o dia inteiro — a diferença
é que eles **leem** o erro em vez de se assustarem com ele.

Vamos provocar um erro de propósito.

````python
# Este código está errado de propósito: a variável não existe.
try:
    print(varivel_com_erro_de_digitacao)
except NameError as erro:
    print("Python levantou um NameError:")
    print(erro)
````

O importante é aprender a **ler de baixo para cima**. A última linha do erro é a que
interessa; ela tem duas partes:

```
NameError: name 'varivel_com_erro_de_digitacao' is not defined
└── o tipo ──┘ └────────── a explicação ─────────────────────┘
```

Os quatro tipos que você mais vai encontrar nas próximas semanas:

| Erro | Tradução prática | Causa típica |
|---|---|---|
| `NameError` | "esse nome não existe" | erro de digitação, ou você não executou a célula que criava a variável |
| `SyntaxError` | "isso não é Python válido" | parêntese ou aspas que não fecharam, `:` esquecido |
| `TypeError` | "essa operação não vale para esse tipo" | somar texto com número |
| `FileNotFoundError` | "não achei esse arquivo" | caminho errado, ou você não clonou o repositório |
| `ModuleNotFoundError` | "essa biblioteca não está aqui" | biblioteca não instalada, ou kernel errado no VS Code |

**Como pedir ajuda direito:** copie a **última linha** do erro e cole no Google, ou no
grupo da capacitação, junto com a célula que você rodou. Isso resolve a enorme maioria
dos casos em segundos. "Deu erro" sem o texto do erro não é uma pergunta respondível.

Diga também **onde** você está rodando — Colab ou VS Code. Boa parte dos problemas de
ambiente tem solução diferente nos dois.

## 9. Recapitulando

- Um **notebook** mistura texto, código e resultado no mesmo documento, e é o formato
  padrão da ciência de dados porque análise de dados é um processo de idas e vindas.
- Células de código rodam com `Shift + Enter`; o número `[n]` mostra a **ordem de
  execução**, não a posição na tela.
- O notebook guarda **estado**: variáveis criadas em uma célula sobrevivem para as
  outras. Antes de entregar qualquer coisa, reinicie e execute tudo.
- Você tem dois ambientes possíveis, e **um só já basta**:
  - **Colab** — recomendado, zero instalação, roda no navegador; lembre de salvar a cópia
    no Drive e de rodar a célula de *setup* dos notebooks com dados a cada sessão.
  - **VS Code** — local, exige Python, a extensão **Jupyter**, um `venv` e o
    `pip install -r requirements.txt`; em troca, acesso direto aos arquivos do
    repositório e funcionamento offline.
- No VS Code, o **kernel** é o Python que executa o notebook. Erro de biblioteca que
  "deveria estar instalada" é, quase sempre, kernel errado.
- Bibliotecas se importam com apelidos convencionais: `pd`, `np`, `plt`, `sns`.
- Erro é rotina. Leia a **última linha**, identifique o tipo, pesquise a mensagem.
