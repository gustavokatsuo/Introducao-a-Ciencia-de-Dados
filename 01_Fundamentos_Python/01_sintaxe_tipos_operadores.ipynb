## 1. Variáveis

Uma variável é um nome que aponta para um valor. Você cria uma com o sinal de igual:

```python
nome = valor
```

Leia o `=` como **"recebe"**, não como "é igual a". `preco = 32.50` significa
"a caixinha chamada `preco` passa a guardar o valor 32.50".

````python
ticker = "PETR4"
preco_fechamento = 32.50
quantidade = 100

print(ticker)
print(preco_fechamento)
print(quantidade)
````

Variáveis existem para serem reutilizadas e recombinadas:

````python
valor_total = preco_fechamento * quantidade
print(valor_total)
````

### Regras e convenções de nomes

**Regras** (se você quebrar, o Python recusa):

- só letras, números e `_`;
- não pode começar com número (`3m` é erro, `m3` é válido);
- não pode ser uma palavra reservada da linguagem (`if`, `for`, `class`, `return`...).

**Convenções** (o Python aceita, mas a comunidade vai torcer o nariz):

- use `snake_case`: tudo minúsculo, palavras separadas por `_` →
  `preco_medio`, `taxa_de_juros`;
- use nomes **descritivos**. `x` não diz nada; `retorno_anual` diz tudo;
- escreva em uma língua só. Neste material usamos português para os nomes que nós
  criamos, e mantemos o inglês do que vem das bibliotecas.

| Evite | Prefira |
|---|---|
| `x = 32.5` | `preco = 32.5` |
| `precoMedioDaAcao` | `preco_medio_acao` |
| `lista1`, `lista2` | `precos_petr4`, `precos_vale3` |
| `df2` | `acoes_filtradas` |

Isso não é preciosismo. Você vai reler seu próprio código daqui a duas semanas sem
lembrar de nada — nomes bons são a diferença entre reler e reescrever.

### Comentários

Tudo depois de `#` na linha é ignorado pelo Python. Comentários servem para explicar
**por que** o código faz algo, não **o que** ele faz (isso o próprio código já diz).

````python
# Corretagem fixa da corretora (tabela de 2025)
custo_corretagem = 4.90

valor_liquido = valor_total - custo_corretagem   # desconta o custo da operação
print(valor_liquido)
````

## 2. Os tipos básicos

Todo valor em Python tem um **tipo**, e o tipo determina o que dá para fazer com ele.
Somar dois números dá um número; "somar" dois textos gruda um no outro. Mesmo símbolo,
comportamentos diferentes — por isso o tipo importa.

| Tipo | Nome | O que guarda | Exemplo |
|---|---|---|---|
| `int` | inteiro | números sem casa decimal | `100`, `-3`, `0` |
| `float` | ponto flutuante | números com casa decimal | `32.50`, `-0.015` |
| `str` | string | texto | `"PETR4"`, `'Ana'` |
| `bool` | booleano | verdadeiro ou falso | `True`, `False` |
| `NoneType` | nulo | ausência de valor | `None` |

A função `type()` responde qual é o tipo de qualquer coisa.

````python
print(type(100))
print(type(32.50))
print(type("PETR4"))
print(type(True))
print(type(None))
````

### int e float

Em Python, `10` e `10.0` são valores diferentes: um é `int`, outro é `float`. Na prática
o Python converte sozinho quando você mistura os dois, então isso raramente atrapalha.

Uma pegadinha que vale conhecer desde já: **`float` é uma aproximação binária**, então
contas com decimais podem dar resultados levemente estranhos.

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

Não é bug do Python — é como computadores representam decimais, e vale para Excel, R,
Java, qualquer linguagem. As consequências para você:

- **nunca** compare dois `float` com `==`; compare a diferença contra uma tolerância;
- para exibir valores, arredonde: `round(valor, 2)`;
- se um dia trabalhar com dinheiro em sistema transacional (contabilidade, pagamentos),
  use tipos decimais próprios. Para análise de dados, `float` é o padrão e está ótimo.

````python
resultado = 0.1 + 0.2

print(round(resultado, 2))
print(abs(resultado - 0.3) < 1e-9)   # jeito correto de comparar floats
````

### str (texto)

Strings ficam entre aspas — simples ou duplas, tanto faz, desde que abra e feche com a
mesma. Escolha uma e seja consistente.

````python
empresa = "Petróleo Brasileiro S.A."
setor = 'Petróleo e Gás'

print(empresa)
print(setor)
print(len(empresa))       # len() conta os caracteres
````

Strings têm **métodos**: funções que pertencem ao próprio valor e são chamadas com um
ponto. Estes cinco você vai usar muito na hora de limpar dados:

````python
texto_sujo = "  petr4  "

print(texto_sujo.strip())            # remove espaços das pontas
print(texto_sujo.strip().upper())    # e deixa maiúsculo
print("PETR4".lower())               # minúsculo
print("PETR4".replace("4", "3"))     # substitui
print("PETR4,VALE3,ITUB4".split(","))  # quebra em partes -> vira uma lista
````

> Repare no encadeamento: `texto_sujo.strip().upper()`. O `.strip()` devolve um texto
> novo, e o `.upper()` é aplicado sobre esse resultado. Isso vai aparecer o tempo todo
> quando chegarmos em pandas.

Você também pode acessar pedaços de uma string pela posição. **Em Python a contagem
começa em zero** — o primeiro caractere é o `[0]`.

````python
ticker = "PETR4"

print(ticker[0])      # primeiro caractere
print(ticker[-1])     # último (índices negativos contam de trás para frente)
print(ticker[0:4])    # do índice 0 até o 4, sem incluir o 4
print(ticker[:4])     # mesma coisa, forma abreviada
````

Esse `[inicio:fim]` chama-se **fatiamento** (*slicing*), e a regra "inclui o início,
exclui o fim" vale para toda a linguagem. Você vai reencontrá-la em listas e em pandas.

### bool (verdadeiro ou falso)

Só dois valores possíveis: `True` e `False` — com maiúscula, sempre. Eles são o
resultado natural de qualquer comparação, e é com eles que vamos filtrar dados no
módulo 02.

````python
preco_atual = 32.50
preco_alvo = 30.00

print(preco_atual > preco_alvo)
print(type(preco_atual > preco_alvo))
````

### None (ausência de valor)

`None` representa "não há valor aqui". Não é zero, não é texto vazio: é a ausência.

Isso é mais relevante do que parece. Bases reais são cheias de buracos — o cliente que
não informou a renda, o dia em que a bolsa não abriu. Em pandas esses buracos aparecem
como `NaN`, e vamos dedicar uma aula inteira a eles.

````python
patrimonio_informado = None

print(patrimonio_informado)
print(patrimonio_informado is None)   # o jeito correto de testar: "is None"
````

## 3. Operadores aritméticos

| Operador | Operação | Exemplo | Resultado |
|---|---|---|---|
| `+` | soma | `7 + 2` | `9` |
| `-` | subtração | `7 - 2` | `5` |
| `*` | multiplicação | `7 * 2` | `14` |
| `/` | divisão | `7 / 2` | `3.5` |
| `//` | divisão inteira | `7 // 2` | `3` |
| `%` | resto da divisão | `7 % 2` | `1` |
| `**` | potência | `7 ** 2` | `49` |

````python
print(7 + 2)
print(7 / 2)      # divisão SEMPRE devolve float, mesmo quando dá exato
print(7 // 2)     # descarta a parte decimal
print(7 % 2)      # sobra
print(7 ** 2)
````

`//` e `%` parecem curiosidades, mas resolvem problemas concretos: "quantas ações
inteiras eu compro com R$ 1.000?" é `//`, e "esse número é par?" é `% 2 == 0`.

````python
capital = 1000.00
preco_acao = 32.50

acoes_compradas = capital // preco_acao
sobra = capital % preco_acao

print("Ações compradas:", acoes_compradas)
print("Sobra em caixa:", round(sobra, 2))
````

### Precedência

Python segue a ordem matemática usual: `**` primeiro, depois `*`, `/`, `//`, `%`, e por
último `+` e `-`. Parênteses mandam em tudo.

Na dúvida, **use parênteses**. Eles custam dois caracteres e evitam bugs silenciosos.

````python
print(2 + 3 * 4)      # 14: a multiplicação vem antes
print((2 + 3) * 4)    # 20: os parênteses mudam a ordem
````

### Aplicação: juros compostos

A fórmula do montante é $M = C \times (1 + i)^t$. Em Python, isso é quase literal:

````python
capital_inicial = 10_000.00     # o _ é só para legibilidade: 10_000 == 10000
taxa_mensal = 0.008             # 0,8% ao mês
meses = 12

montante = capital_inicial * (1 + taxa_mensal) ** meses
juros_ganhos = montante - capital_inicial

print("Montante após 12 meses:", round(montante, 2))
print("Juros ganhos:", round(juros_ganhos, 2))
````

### Operadores de atribuição composta

`x += 1` é um atalho para `x = x + 1`. Existe o equivalente para todas as operações
(`-=`, `*=`, `/=`). Aparece muito dentro de loops, na próxima aula.

````python
saldo = 1000.0

saldo += 250.0    # aporte
saldo -= 80.0     # taxa
saldo *= 1.01     # rendimento de 1%

print(round(saldo, 2))
````

## 4. Operadores de comparação

Todos devolvem `True` ou `False`.

| Operador | Significado |
|---|---|
| `==` | igual a |
| `!=` | diferente de |
| `>` `<` | maior / menor |
| `>=` `<=` | maior ou igual / menor ou igual |

> **Atenção — `=` atribui, `==` compara.** Confundir os dois é o erro mais comum de quem
> está começando. `preco = 30` muda o valor; `preco == 30` pergunta se ele é 30.

````python
preco = 32.50

print(preco == 32.50)
print(preco != 30.00)
print(preco >= 32.50)
print("PETR4" == "petr4")    # texto diferencia maiúscula de minúscula!
````

Aquele último resultado é uma fonte clássica de bug em dados reais: `"São Paulo"`,
`"SÃO PAULO"` e `"sao paulo"` são três valores distintos para o Python. É exatamente por
isso que a limpeza de dados quase sempre passa por padronizar texto com `.lower()` e
`.strip()`.

## 5. Operadores lógicos

Servem para combinar condições:

- `and` → verdadeiro só se **ambos** os lados forem verdadeiros;
- `or` → verdadeiro se **pelo menos um** lado for verdadeiro;
- `not` → inverte.

````python
preco = 32.50
volume = 45_000_000

barato = preco < 35.00
liquido = volume > 10_000_000

print("Barato e líquido:", barato and liquido)
print("Barato ou líquido:", barato or liquido)
print("Não é barato:", not barato)
````

Um recurso elegante do Python: comparações podem ser encadeadas como na matemática.

````python
idade = 27

print(18 <= idade <= 65)              # em Python isso funciona
print(idade >= 18 and idade <= 65)    # equivalente, mais verboso
````

## 6. Conversão de tipos

Misturar tipos incompatíveis gera erro. Este é `TypeError`, seu futuro conhecido:

````python
try:
    resultado = "100" + 50
except TypeError as erro:
    print("TypeError:", erro)
````

`"100"` é texto e `50` é número; o Python se recusa a adivinhar o que você queria. A
solução é converter explicitamente com `int()`, `float()` ou `str()`.

````python
print(int("100") + 50)         # texto -> número
print("100" + str(50))         # número -> texto (concatena!)
print(float("32.50") * 2)
print(int(32.99))              # trunca, NÃO arredonda
print(round(32.99))            # arredonda de verdade
````

> **Atenção:** `int(32.99)` devolve `32`. Ele **corta** a parte decimal em vez de
> arredondar. Para arredondar, use `round()`.

Isso é rotina em dados reais: planilhas frequentemente trazem números salvos como texto
(`"1.250,00"`, `"R$ 4.900"`). Você vai reencontrar esse problema — e resolvê-lo em
escala — na aula de limpeza de dados.

````python
# Nem toda conversão é possível:
try:
    int("trinta e dois")
except ValueError as erro:
    print("ValueError:", erro)
````

## 7. f-strings: montando textos

Para juntar texto e variáveis, use **f-string**: um `f` antes das aspas permite colocar
variáveis entre chaves `{}`.

````python
ticker = "PETR4"
preco = 32.4567
variacao = 0.0231

print(f"{ticker} fechou a R$ {preco}")
````

E dá para formatar o número dentro das próprias chaves — o que torna qualquer relatório
imediatamente mais apresentável:

````python
print(f"Preço: R$ {preco:.2f}")          # 2 casas decimais
print(f"Variação: {variacao:.2%}")       # como porcentagem
print(f"Volume: {45_000_000:,}")         # separador de milhar
print(f"{ticker:>10} | {preco:>8.2f}")   # alinhado à direita, útil em tabelas
````

````python
# Um mini-relatório, juntando tudo desta aula:
quantidade = 300
custo_unitario = 30.10
preco_atual = 32.45

investido = quantidade * custo_unitario
valor_hoje = quantidade * preco_atual
lucro = valor_hoje - investido
retorno = lucro / investido

print(f"Posição em {ticker}")
print(f"  Investido : R$ {investido:>10,.2f}")
print(f"  Hoje      : R$ {valor_hoje:>10,.2f}")
print(f"  Resultado : R$ {lucro:>10,.2f}  ({retorno:.2%})")
````
