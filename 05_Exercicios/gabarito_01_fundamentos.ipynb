## Exercício 1 (teoria) — `=` e `==`


Explique a diferença entre `=` e `==` em Python. Depois, diga o que cada linha abaixo faz
(ou por que falha):

```python
preco = 32.50
preco == 32.50
preco = "trinta e dois"
preco + 1
```

**Resposta:**


`=` é o operador de **atribuição**: cria (ou atualiza) uma variável, guardando um valor
nela. `==` é o operador de **comparação**: pergunta se dois valores são iguais e devolve
`True` ou `False`.

Linha a linha:

1. `preco = 32.50` — cria a variável `preco` com o valor 32.5 (um `float`).
2. `preco == 32.50` — compara e devolve `True`. Não altera nada.
3. `preco = "trinta e dois"` — **sobrescreve** `preco`, que agora guarda um texto (`str`).
   Python permite trocar o tipo de uma variável livremente.
4. `preco + 1` — levanta `TypeError`: não é possível somar `str` com `int`. O Python se
   recusa a adivinhar o que você queria.

## Exercício 2 (código) — Relatório de uma posição


Crie variáveis para uma posição em ações: ticker (`"VALE3"`), quantidade (`150`), preço de
compra (`68.40`) e preço atual (`61.20`).

Calcule o valor investido, o valor atual, o resultado em reais e o retorno percentual.
Imprima um relatório usando **f-strings**, com os valores monetários em duas casas
decimais e o retorno formatado como porcentagem.


> **Dica:** `f"{valor:.2f}"` formata com duas casas; `f"{retorno:.2%}"` formata como porcentagem.

````python
ticker = "VALE3"
quantidade = 150
preco_compra = 68.40
preco_atual = 61.20

valor_investido = quantidade * preco_compra
valor_atual = quantidade * preco_atual
resultado = valor_atual - valor_investido
retorno = resultado / valor_investido

print(f"Posição em {ticker}")
print(f"  Quantidade : {quantidade}")
print(f"  Investido  : R$ {valor_investido:,.2f}")
print(f"  Valor atual: R$ {valor_atual:,.2f}")
print(f"  Resultado  : R$ {resultado:,.2f}")
print(f"  Retorno    : {retorno:.2%}")
````

Note que o retorno é calculado sobre o **valor investido**, não sobre o preço. Como a
quantidade aparece nos dois lados da divisão, ela se cancela — o retorno percentual seria
o mesmo calculando só com os preços. Vale conferir isso mentalmente sempre que um
resultado percentual parecer estranho.

## Exercício 3 (código) — De texto para número


Uma planilha exportou o valor abaixo como texto:

```python
valor_texto = "R$ 1.234,56"
```

Converta-o para um `float` usando apenas métodos de string e a função `float()`. Imprima
o resultado multiplicado por 2 para confirmar que virou número.


> **Dica:** Você precisa remover `"R$ "`, remover o ponto de milhar e trocar a vírgula por ponto. A ordem dessas operações importa.

````python
valor_texto = "R$ 1.234,56"

limpo = valor_texto.replace("R$", "").strip()   # tira o símbolo e os espaços
limpo = limpo.replace(".", "")                  # tira o ponto de milhar
limpo = limpo.replace(",", ".")                 # vírgula decimal vira ponto

valor = float(limpo)

print(limpo)
print(valor * 2)
print(type(valor))
````

**A ordem importa.** Se você trocasse a vírgula por ponto antes de remover o ponto de
milhar, `"1.234,56"` viraria `"1.234.56"` — que `float()` não consegue converter.

Este exercício é uma prévia de um problema que você vai reencontrar em escala no módulo
02: colunas inteiras de valores monetários salvas como texto.

## Exercício 4 (teoria) — Aritmética de ponto flutuante


Execute mentalmente (e depois no notebook, se quiser):

```python
0.1 + 0.2 == 0.3
```

O resultado é `False`. Explique por quê e diga como comparar dois números decimais
corretamente.

**Resposta:**


Computadores guardam números decimais em **base 2**, e frações como 0,1 e 0,2 não têm
representação binária exata — assim como 1/3 não tem representação decimal exata. O que
fica na memória são aproximações muito próximas, e a soma delas resulta em algo como
`0.30000000000000004`, que não é exatamente `0.3`.

Isso não é um defeito do Python: vale para qualquer linguagem e também para o Excel.

Como comparar corretamente: verifique se a **diferença** entre os dois valores é menor que
uma tolerância pequena.

```python
abs((0.1 + 0.2) - 0.3) < 1e-9    # True
```

Para exibir valores, use `round(valor, 2)`. E, se um dia trabalhar com dinheiro em um
sistema transacional (contabilidade, pagamentos), use tipos decimais próprios em vez de
`float`.

## Exercício 5 (código) — Classificando a variação do dia


Escreva uma cadeia `if` / `elif` / `else` que classifique a variação diária de uma ação:

| Variação | Classificação |
|---|---|
| acima de +3% | `"Alta forte"` |
| entre 0% e +3% | `"Alta"` |
| exatamente 0% | `"Estável"` |
| entre 0% e −3% | `"Queda"` |
| abaixo de −3% | `"Queda forte"` |

Teste com os valores `0.045`, `0.012`, `0.0`, `-0.008` e `-0.052`.


> **Dica:** Use um `for` para percorrer a lista de valores de teste, em vez de repetir a cadeia cinco vezes.

````python
variacoes = [0.045, 0.012, 0.0, -0.008, -0.052]

for variacao in variacoes:
    if variacao > 0.03:
        classificacao = "Alta forte"
    elif variacao > 0:
        classificacao = "Alta"
    elif variacao == 0:
        classificacao = "Estável"
    elif variacao > -0.03:
        classificacao = "Queda"
    else:
        classificacao = "Queda forte"

    print(f"{variacao:>7.2%} -> {classificacao}")
````

A **ordem das condições** é o ponto do exercício. Se `elif variacao > 0` viesse antes de
`if variacao > 0.03`, nenhuma variação seria classificada como "Alta forte", porque a
condição mais geral capturaria todas primeiro. Regra: do mais específico para o mais
geral.

## Exercício 6 (código) — Quantos meses até a meta?


Você tem R$ 5.000 aplicados a 0,9% ao mês e faz aportes de R$ 300 no fim de cada mês.

Use um `while` para descobrir **quantos meses** levam até o saldo alcançar R$ 20.000.
Imprima o número de meses e o saldo final.


> **Dica:** A cada volta: renda o saldo (`saldo *= 1 + taxa`), some o aporte e conte mais um mês.

````python
saldo = 5_000.00
taxa = 0.009
aporte = 300.00
meta = 20_000.00

meses = 0
while saldo < meta:
    saldo = saldo * (1 + taxa) + aporte
    meses += 1

print(f"Meta atingida em {meses} meses ({meses / 12:.1f} anos)")
print(f"Saldo final: R$ {saldo:,.2f}")
````

Todo `while` precisa de algo que faça a condição eventualmente ficar falsa — aqui, o
saldo crescendo. Se o aporte fosse zero **e** a taxa fosse zero, o laço nunca terminaria.
Ao escrever um `while`, sempre pergunte: *o que garante que isso acaba?*

## Exercício 7 (código) — Relatório com `zip`


Dadas as três listas abaixo, imprima uma linha por ativo com ticker, quantidade, preço e
valor da posição, alinhados em colunas. No fim, imprima o **valor total** da carteira.

Use `zip` — não use índices.


> **Dica:** `f"{texto:<8}"` alinha à esquerda em 8 caracteres; `f"{numero:>10.2f}"` alinha à direita.

````python
tickers = ["PETR4", "VALE3", "ITUB4", "WEGE3"]
quantidades = [100, 50, 200, 80]
precos = [32.50, 61.20, 28.75, 41.90]

total = 0

print(f"{'ATIVO':<8}{'QTD':>6}{'PREÇO':>10}{'VALOR':>14}")
print("-" * 38)

for ticker, quantidade, preco in zip(tickers, quantidades, precos):
    valor = quantidade * preco
    total += valor
    print(f"{ticker:<8}{quantidade:>6}{preco:>10.2f}{valor:>14,.2f}")

print("-" * 38)
print(f"{'TOTAL':<8}{'':>16}{total:>14,.2f}")
````

`zip` para na lista mais curta, **sem avisar**. Se as três listas viessem de fontes
diferentes, valeria conferir antes:

```python
assert len(tickers) == len(quantidades) == len(precos)
```

## Exercício 8 (código) — Manipulando uma lista


Partindo de `carteira = ["PETR4", "VALE3", "ITUB4"]`, faça, nesta ordem:

1. adicione `"WEGE3"` ao fim;
2. adicione `"ABEV3"` na primeira posição;
3. remova `"VALE3"`;
4. imprima a lista ordenada alfabeticamente **sem** alterar a original;
5. imprima quantos ativos restaram.

Imprima a lista após cada passo.


> **Dica:** `sorted(lista)` devolve uma cópia ordenada; `lista.sort()` altera a original.

````python
carteira = ["PETR4", "VALE3", "ITUB4"]
print("inicial :", carteira)

carteira.append("WEGE3")
print("append  :", carteira)

carteira.insert(0, "ABEV3")
print("insert  :", carteira)

carteira.remove("VALE3")
print("remove  :", carteira)

print("ordenada:", sorted(carteira))
print("original:", carteira, "(intacta)")
print("total   :", len(carteira), "ativos")
````

## Exercício 10 (teoria) — Lista ou dicionário?


Para cada situação, diga qual estrutura você usaria (lista, dicionário, tupla ou
conjunto) e justifique em uma frase:

1. Os preços de fechamento de uma ação nos últimos 250 pregões.
2. As características de uma empresa: nome, setor, ano de fundação, CNPJ.
3. Os setores distintos presentes em uma carteira.
4. As coordenadas de um ponto que não devem ser alteradas.
5. O total investido em cada ativo, consultado pelo ticker.

**Resposta:**


1. **Lista** — é uma sequência ordenada de valores do mesmo tipo, e a ordem (cronológica)
   é essencial.
2. **Dicionário** — cada valor tem um nome próprio; `empresa["setor"]` se explica

3. **Conjunto** — a pergunta é exatamente "quais são os valores distintos", que é a

4. **Tupla** — imutável por construção, protegendo contra alteração acidental.
5. **Dicionário** — a consulta é por chave (`totais["PETR4"]`), que é o caso de uso
   central do dicionário.

## Exercício 11 (código) — Carteira como dicionário


Dado o dicionário `posicoes`, que mapeia ticker para valor investido:

1. imprima o valor total investido;
2. imprima o ticker da **maior** posição;
3. imprima cada ativo com sua participação percentual na carteira, em ordem decrescente
   de valor.


> **Dica:** `max(dicionario, key=dicionario.get)` devolve a chave com maior valor. Para ordenar, `sorted(dicionario.items(), key=lambda par: par[1], reverse=True)`.

````python
posicoes = {"PETR4": 3250.00, "VALE3": 3060.00, "ITUB4": 5750.00, "WEGE3": 3352.00}

total = sum(posicoes.values())
print(f"Total investido: R$ {total:,.2f}")

maior = max(posicoes, key=posicoes.get)
print(f"Maior posição: {maior} (R$ {posicoes[maior]:,.2f})")

print("\nParticipação:")
for ticker, valor in sorted(posicoes.items(), key=lambda par: par[1], reverse=True):
    print(f"  {ticker}: {valor / total:>6.1%}")
````

`max(posicoes, key=posicoes.get)` merece atenção: sem o `key`, `max` compararia as
**chaves** (os textos) e devolveria `"WEGE3"` por ordem alfabética. O `key` diz por qual
critério comparar.

## Exercício 13 (teoria) — `return` e `print`


Observe as duas funções:

```python
def dobro_a(x):
    return x * 2

def dobro_b(x):
    print(x * 2)
```

1. O que aparece na tela ao executar `dobro_a(5)` em uma célula? E `dobro_b(5)`?
2. O que acontece com `resultado = dobro_b(5)` e depois `resultado + 1`?
3. Qual das duas você usaria numa função que calcula um retorno? Por quê?

**Resposta:**


1. `dobro_a(5)` **devolve** 10. Em um notebook, o valor da última linha da célula é
   exibido automaticamente, então você vê `10` — mas quem exibiu foi o notebook, não a
   função. `dobro_b(5)` **imprime** 10 na tela e devolve `None`.

2. `resultado` fica com `None` (a função não devolveu nada). `resultado + 1` levanta
   `TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'`.

3. **`return`**, sem dúvida. Uma função que calcula deve **entregar** o valor para quem
   chamou, para que ele possa ser guardado, somado, colocado numa tabela ou num gráfico.
   Quem chama decide se quer exibir — e faz isso com `print(calcular_retorno(...))`.
   Funções que só imprimem são um beco sem saída: o número aparece e se perde.

## Exercício 14 (código) — Função com valor padrão


Escreva a função `calcular_montante(capital, taxa, periodos=12)` que devolve o montante
de juros compostos, com uma **docstring** explicando o que ela faz, o que recebe e o que
devolve.

Teste-a três vezes: usando o padrão, passando 24 períodos por posição e passando 60 por
nome.

````python
def calcular_montante(capital, taxa, periodos=12):
    """
    Calcula o montante de uma aplicação a juros compostos.

    Parâmetros
    ----------
    capital : float
        Valor aplicado inicialmente.
    taxa : float
        Taxa por período, em decimal (0.008 = 0,8%).
    periodos : int, opcional
        Número de períodos. O padrão é 12.

    Retorna
    -------
    float
        Montante ao final dos períodos.
    """
    return capital * (1 + taxa) ** periodos


print(round(calcular_montante(10_000, 0.008), 2))
print(round(calcular_montante(10_000, 0.008, 24), 2))
print(round(calcular_montante(10_000, 0.008, periodos=60), 2))

help(calcular_montante)
````

Parâmetros com valor padrão precisam vir **depois** dos obrigatórios. Passar argumentos
por nome (`periodos=60`) deixa a chamada muito mais legível quando há vários parâmetros —
e protege contra trocar a ordem sem perceber.

## Exercício 15 (código) — Função com vários retornos


Escreva a função `resumir(valores)` que recebe uma lista de números e devolve **quatro**
valores: mínimo, máximo, média e amplitude (máximo − mínimo).

Chame a função desempacotando o resultado em quatro variáveis e imprima um resumo
formatado.


> **Dica:** Basta separar os valores por vírgula no `return`.

````python
def resumir(valores):
    """Devolve mínimo, máximo, média e amplitude de uma lista de números."""
    minimo = min(valores)
    maximo = max(valores)
    media = sum(valores) / len(valores)
    return minimo, maximo, media, maximo - minimo


precos = [32.50, 61.20, 28.75, 41.90, 15.30]
minimo, maximo, media, amplitude = resumir(precos)

print(f"Mínimo   : R$ {minimo:>7.2f}")
print(f"Máximo   : R$ {maximo:>7.2f}")
print(f"Média    : R$ {media:>7.2f}")
print(f"Amplitude: R$ {amplitude:>7.2f}")
````

Por baixo, `return a, b, c, d` devolve uma **tupla**. O desempacotamento
(`minimo, maximo, media, amplitude = ...`) distribui os elementos em variáveis, na ordem.
Se você escrever menos variáveis que valores, o Python levanta um `ValueError`.

## Exercício 16 (código) — Desafio — análise completa de carteira


Junte tudo. Escreva **três funções**:

1. `calcular_retorno(preco_compra, preco_atual)` — devolve o retorno em decimal;
2. `classificar(retorno, limite=0.10)` — devolve `"Lucro relevante"`, `"Lucro pequeno"`,
   `"Perda pequena"` ou `"Perda relevante"`;
3. `analisar(carteira)` — recebe a lista de dicionários, devolve **duas** coisas: uma
   lista de dicionários com ticker, valor atual, retorno e classificação, e o valor total
   da carteira.

Depois, imprima um relatório com uma linha por ativo (incluindo o peso de cada posição na
carteira) e uma linha de total.


> **Dica:** Escreva e teste uma função por vez, antes de juntar tudo.

````python
def calcular_retorno(preco_compra, preco_atual):
    """Retorno percentual entre o preço de compra e o atual, em decimal."""
    return (preco_atual - preco_compra) / preco_compra


def classificar(retorno, limite=0.10):
    """Classifica um retorno usando um limite de relevância."""
    if retorno > limite:
        return "Lucro relevante"
    if retorno > 0:
        return "Lucro pequeno"
    if retorno > -limite:
        return "Perda pequena"
    return "Perda relevante"


def analisar(carteira):
    """
    Analisa uma carteira de ações.

    Parâmetros
    ----------
    carteira : list[dict]
        Cada item precisa ter ticker, preco_compra, preco_atual e quantidade.

    Retorna
    -------
    tuple[list[dict], float]
        A análise de cada posição e o valor total atual da carteira.
    """
    resultado = []
    total = 0.0

    for posicao in carteira:
        valor = posicao["preco_atual"] * posicao["quantidade"]
        retorno = calcular_retorno(posicao["preco_compra"], posicao["preco_atual"])
        total += valor
        resultado.append({
            "ticker": posicao["ticker"],
            "valor": valor,
            "retorno": retorno,
            "classificacao": classificar(retorno),
        })

    return resultado, total


carteira = [
    {"ticker": "PETR4", "preco_compra": 30.10, "preco_atual": 32.45, "quantidade": 100},
    {"ticker": "VALE3", "preco_compra": 68.40, "preco_atual": 61.20, "quantidade": 50},
    {"ticker": "ITUB4", "preco_compra": 26.90, "preco_atual": 28.75, "quantidade": 200},
    {"ticker": "MGLU3", "preco_compra": 15.60, "preco_atual": 9.85, "quantidade": 300},
]

analise, total = analisar(carteira)

print(f"{'ATIVO':<8}{'VALOR':>12}{'PESO':>9}{'RETORNO':>11}   CLASSIFICAÇÃO")
print("-" * 62)
for linha in analise:
    peso = linha["valor"] / total
    print(f"{linha['ticker']:<8}{linha['valor']:>12,.2f}{peso:>9.1%}"
          f"{linha['retorno']:>11.2%}   {linha['classificacao']}")
print("-" * 62)
print(f"{'TOTAL':<8}{total:>12,.2f}{1:>9.1%}")
````

Repare em três decisões de projeto:

1. **Cada função faz uma coisa só.** `calcular_retorno` não classifica; `classificar` não
   calcula. Isso torna cada uma testável isoladamente.
2. **Nenhuma delas imprime.** Elas devolvem valores; a impressão fica por conta de quem
   chama. Assim as mesmas funções servem para um relatório na tela, para um arquivo ou
   para um gráfico.
3. **`classificar` tem `limite` como parâmetro com padrão.** Se amanhã a política mudar
   para 15%, basta passar `classificar(retorno, limite=0.15)` — sem tocar na função.
