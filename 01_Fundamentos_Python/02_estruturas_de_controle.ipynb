## 1. `if`: tomando decisões

A estrutura é:

```python
if condicao:
    # este bloco roda se a condição for True
```

Duas coisas obrigatórias: os **dois-pontos** no fim da linha do `if`, e a
**indentação** (recuo) do bloco de baixo.

````python
preco = 28.90
preco_alvo = 30.00

if preco < preco_alvo:
    print("Preço abaixo do alvo — avaliar compra")
````

### `else` e `elif`

- `else` → o que fazer quando a condição do `if` é falsa;
- `elif` (*else if*) → testar outra condição, quando a anterior falhou.

````python
variacao = -0.021

if variacao > 0:
    print("Alta no dia")
else:
    print("Queda no dia")
````

````python
variacao = 0.035

if variacao > 0.02:
    classificacao = "Alta forte"
elif variacao > 0:
    classificacao = "Alta moderada"
elif variacao == 0:
    classificacao = "Estável"
elif variacao > -0.02:
    classificacao = "Queda moderada"
else:
    classificacao = "Queda forte"

print(f"Variação de {variacao:.2%}: {classificacao}")
````

Duas propriedades importantes da cadeia `if/elif/else`:

1. **Só um bloco executa.** O Python testa de cima para baixo e para no primeiro que
   der `True`.
2. **A ordem importa.** Se `elif variacao > 0` viesse antes de `if variacao > 0.02`,
   nenhuma alta jamais seria classificada como "forte" — a primeira condição capturaria
   todas. Sempre coloque as condições mais específicas primeiro.

## 2. Indentação: em Python, espaço em branco é sintaxe

Em muitas linguagens, blocos são delimitados por chaves `{ }` e o recuo é só
organização visual. Em Python, **o recuo é a delimitação**. Errar a indentação não é
feio: é erro de sintaxe.

A convenção universal é **4 espaços** por nível. O Colab e o Jupyter fazem isso
automaticamente quando você aperta Enter depois dos dois-pontos.

````python
patrimonio = 250_000
perfil = "Arrojado"

if patrimonio > 100_000:
    print("Cliente qualificado")          # nível 1: dentro do if
    if perfil == "Arrojado":
        print("  Elegível a fundos de ações")   # nível 2: dentro do if de dentro
    print("Fim da análise do cliente")     # de volta ao nível 1

print("Fim do programa")                   # nível 0: fora de tudo, sempre roda
````

````python
# O que acontece quando a indentação está errada:
codigo_errado = '''
if True:
print("sem recuo")
'''

try:
    exec(codigo_errado)
except IndentationError as erro:
    print("IndentationError:", erro)
````

## 3. `for`: repetir para cada item

O `for` percorre uma coleção e executa o bloco uma vez para cada elemento.

```python
for item in colecao:
    # faz algo com item
```

O nome `item` é escolhido por você — e vale a pena escolher bem: `for ticker in
carteira` se lê como português, `for x in y` não se lê como nada.

````python
carteira = ["PETR4", "VALE3", "ITUB4", "WEGE3"]

for ticker in carteira:
    print(f"Analisando {ticker}...")
````

O padrão mais comum de todos: criar uma variável **acumuladora** antes do loop e
atualizá-la a cada volta.

````python
precos = [32.50, 61.20, 28.75, 41.90]

soma = 0                    # acumulador começa zerado
for preco in precos:
    soma += preco           # a cada volta, soma mais um preço

media = soma / len(precos)

print(f"Soma: R$ {soma:.2f}")
print(f"Média: R$ {media:.2f}")
````

> Em Python já existem `sum()` e `len()` prontos — `sum(precos) / len(precos)` faz o
> mesmo em uma linha. Mostramos o loop porque o **padrão do acumulador** é a base de
> muita coisa. Mas, na vida real, prefira a função pronta: menos código, menos bug.

````python
print(sum(precos) / len(precos))
````

## 4. `range()`: gerando sequências de números

`range()` produz uma sequência de inteiros. É o que você usa quando quer repetir algo
*n* vezes, ou percorrer posições.

| Forma | Produz |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 |

Repare de novo na regra: **inclui o início, exclui o fim**. `range(5)` tem cinco
números, mas o último é 4.

````python
for i in range(5):
    print(i, end=" ")
print()

print(list(range(2, 6)))
print(list(range(0, 10, 2)))
print(list(range(5, 0, -1)))
````

````python
# Simulando a evolução de um investimento, ano a ano:
capital = 10_000.00
taxa_anual = 0.11

for ano in range(1, 6):
    capital = capital * (1 + taxa_anual)
    print(f"Ano {ano}: R$ {capital:>10,.2f}")
````

## 5. `enumerate()`: o item e a sua posição

Às vezes você precisa do valor **e** de onde ele está. A tentação é usar
`range(len(lista))` e indexar na mão — funciona, mas é verboso e propenso a erro.
`enumerate()` entrega os dois de uma vez.

````python
carteira = ["PETR4", "VALE3", "ITUB4", "WEGE3"]

# jeito trabalhoso:
for i in range(len(carteira)):
    print(i, carteira[i])

print("---")

# jeito idiomático:
for posicao, ticker in enumerate(carteira):
    print(posicao, ticker)
````

````python
# start=1 quando o número é para humano lerem, não para indexar:
for posicao, ticker in enumerate(carteira, start=1):
    print(f"{posicao}º ativo da carteira: {ticker}")
````

## 6. `zip()`: percorrer duas listas em paralelo

Quando informações relacionadas estão em listas separadas, `zip()` costura elas e
devolve os pares.

````python
tickers = ["PETR4", "VALE3", "ITUB4"]
precos = [32.50, 61.20, 28.75]
quantidades = [100, 50, 200]

for ticker, preco in zip(tickers, precos):
    print(f"{ticker}: R$ {preco:.2f}")
````

````python
# zip aceita quantas listas você quiser:
valor_total = 0

for ticker, preco, qtd in zip(tickers, precos, quantidades):
    valor = preco * qtd
    valor_total += valor
    print(f"{ticker:>6} | {qtd:>4} × R$ {preco:>6.2f} = R$ {valor:>9,.2f}")

print(f"{'TOTAL':>6} | {'':>4}   {'':>10} R$ {valor_total:>9,.2f}")
````

> **Atenção:** Se as listas tiverem tamanhos diferentes, `zip()` para na mais curta, **sem
> avisar**. Isso pode esconder um erro nos seus dados. Se o alinhamento entre as listas
> for crítico, confira os tamanhos antes.

````python
print(len(tickers) == len(precos) == len(quantidades))
````

## 7. `while`: repetir enquanto uma condição valer

O `for` repete uma vez por item. O `while` repete **enquanto** algo for verdade — você
não sabe de antemão quantas voltas serão.

A pergunta que decide qual usar: *"eu sei quantas repetições preciso?"* Se sim, `for`.
Se a resposta é "até acontecer tal coisa", `while`.

````python
# Quantos meses até dobrar o capital a 0,8% ao mês?
capital = 10_000.00
meta = 20_000.00
taxa = 0.008
meses = 0

while capital < meta:
    capital = capital * (1 + taxa)
    meses += 1

print(f"Meta atingida em {meses} meses ({meses/12:.1f} anos)")
print(f"Capital final: R$ {capital:,.2f}")
````

> **Atenção — Loop infinito.** Se a condição do `while` nunca ficar falsa, o programa
> trava para sempre. Toda vez que escrever um `while`, verifique: *o que, dentro do bloco,
> faz a condição eventualmente virar falsa?* No exemplo acima é o `capital` crescendo.
>
> Se travar no Colab, use `Ambiente de execução → Interromper execução` (ou o botão de
> interromper ao lado da célula).

## 8. `break` e `continue`

- `break` → sai do loop imediatamente;
- `continue` → pula o resto desta volta e vai para a próxima.

````python
precos = [32.50, 31.80, 29.40, 33.10, 27.90]
gatilho = 30.00

for dia, preco in enumerate(precos, start=1):
    if preco < gatilho:
        print(f"Dia {dia}: preço {preco} rompeu o gatilho — encerrando busca")
        break
    print(f"Dia {dia}: preço {preco}, seguindo")
````

````python
patrimonios = [50_000, None, 120_000, None, 89_000]

soma = 0
validos = 0

for valor in patrimonios:
    if valor is None:
        continue          # pula os registros sem informação
    soma += valor
    validos += 1

print(f"{validos} registros válidos de {len(patrimonios)}")
print(f"Patrimônio médio: R$ {soma/validos:,.2f}")
````

Esse último exemplo é um retrato fiel do que você vai fazer com dados reais: parte dos
registros está faltando, e você precisa decidir explicitamente o que fazer com eles.
Aqui, ignoramos. Nem sempre essa é a decisão certa — e o módulo 02 discute as
alternativas.

## 9. Juntando tudo

Um mini-classificador de carteira, usando praticamente tudo desta aula.

````python
tickers = ["PETR4", "VALE3", "ITUB4", "WEGE3", "MGLU3"]
precos_compra = [30.10, 68.40, 26.90, 38.20, 15.60]
precos_atuais = [32.45, 61.20, 28.75, 41.90, 9.85]

ganhadores = 0
perdedores = 0

print(f"{'ATIVO':<8}{'COMPRA':>10}{'ATUAL':>10}{'RETORNO':>12}  SITUAÇÃO")
print("-" * 55)

for ticker, compra, atual in zip(tickers, precos_compra, precos_atuais):
    retorno = (atual - compra) / compra

    if retorno > 0.10:
        situacao = "Lucro relevante"
        ganhadores += 1
    elif retorno > 0:
        situacao = "Lucro pequeno"
        ganhadores += 1
    elif retorno > -0.10:
        situacao = "Perda pequena"
        perdedores += 1
    else:
        situacao = "Perda relevante"
        perdedores += 1

    print(f"{ticker:<8}{compra:>10.2f}{atual:>10.2f}{retorno:>11.2%}  {situacao}")

print("-" * 55)
print(f"{ganhadores} ativos no lucro, {perdedores} no prejuízo")
````

## 10. Recapitulando

- `if` / `elif` / `else` decidem o caminho. Só um bloco executa, e a **ordem das
  condições importa** — do mais específico para o mais geral.
- **Indentação é sintaxe.** Quatro espaços por nível, sempre.
- `for` percorre uma coleção; `while` repete enquanto a condição for verdadeira. Todo
  `while` precisa de algo que o faça terminar.
- `range(inicio, fim, passo)` gera números; exclui o fim.
- `enumerate()` dá posição + valor; `zip()` percorre listas em paralelo. Prefira os dois
  a `range(len(...))`.
- `break` sai do loop; `continue` pula para a próxima volta.
